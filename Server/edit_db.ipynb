```python
import sqlite3

# Path to database file
DATABASE = 'server_database.db'

# Create Tables
conn = sqlite3.connect(DATABASE)
cursor = conn.cursor()
```

```python
# Enable foreign key constraints
cursor.execute("PRAGMA foreign_keys = ON;")

# Create devices table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS devices (
        id TEXT PRIMARY KEY,
        key_LSB INTEGER,
        create_time DATETIME DEFAULT CURRENT_TIMESTAMP,
        last_event_time DATETIME, 
        last_live_time DATETIME,
        config_file TEXT -- Use TEXT to store JSON data
    )
''')

print("Database and tables created successfully.")

conn.close()
```

```python


cursor.execute('''
    INSERT INTO devices (id, key_LSB, last_event_time, last_live_time, config_file) 
    VALUES (?, ?, ?, ?, ?)
''', (
    'device_002',            # Device ID
     2365,                   # Key LSB
    '2024-12-30 12:30:00',   # Last event time
    '2024-12-30 12:35:00',   # Last live time
    '{"setting": "value2"}'  # Config file (stored as JSON string)
))

conn.commit()
cursor.execute("SELECT * FROM config")
print("Config Table:", cursor.fetchall())
conn.close()
```
